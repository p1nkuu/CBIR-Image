```python
# loading in pubmed dataset json file
nlp = spacy.load("en_core_web_sm")

with open('/Users/aditibelle/Desktop/pubmed_set/captions.json', 'r') as file:
    pubmed_data = json.loads(file.read())
```

```python
# compiling all the captions into one string
captions = ''

for id in pubmed_data:
    if pubmed_data[id]['caption'] not in captions:
        captions += pubmed_data[id]['caption']

captions = captions.strip()
```

```python
doc = nlp(captions)

filtered_tokens = [token.text for token in doc if not token.is_stop and not token.is_punct]
print(filtered_tokens)
```

```python
# match up the captions with their ids in a dictionary
caption_ids = {}

for id in pubmed_data:
    if pubmed_data[id]['uuid'] not in caption_ids.keys():
        caption_ids[pubmed_data[id]['uuid']] = pubmed_data[id]['caption'].strip()
caption_ids
```

# Sort Images by Common Staining Techniques

Used https://www.ncbi.nlm.nih.gov/books/NBK557663/ as a reference for common types of stains.

```python
stain_tech = ['giemsa', 'lfb', 'lfb-pas', 'periodic acid-schiff', 'pas-positive', 'masson\'s trichrome', 'massons trichrome', 'masson trichrome', 'masson fontana',
              'papanicolaou', 'hematoxylin and eosin', 'h&e', 'ub immunostaining', 'ihc', 'immunohistochemical staining', 
              ]

grouped_captions = {keyword.lower(): [] for keyword in stain_tech}

for uuid in caption_ids:
    caption_lower = caption_ids[uuid].lower()
    for keyword in stain_tech:
        if keyword in caption_lower:
            grouped_captions[keyword].append(uuid)

# Print grouped captions
for keyword, group in grouped_captions.items():
    print(f"Captions containing '{keyword}':")
    for caption in group:
        print(f" - {caption}")
    print(f"Number of images with {keyword}: " + str(len(grouped_captions[keyword])) + "\n")
```

# Sort Images by Biological Stains

```python
bio_stains = ['bismarck brown', 'carmine', 'coomassie blue', 'cresyl violet', 'crystal violet', 'dapi', 'eosin', 
                       'ethidium bromide', 'acid fuchsin', 'haematoxylin', 'hoechst', 'iodine', 'malachite green', 'methyl green',
                       'methylene blue', 'neutral red', 'nile blue', 'nile red', 'osmium tetroxide', 'osmium tetraoxide', 'propidium iodide',
                       'rhodamine', 'safranine', 'prussian blue', 'alcian blue', 'congo red', 'acridine orange', 'silver', 'mucicarmine']

grouped_captions = {keyword.lower(): [] for keyword in bio_stains}

# Grouping the captions based on the presence of keywords
for uuid in caption_ids:
    caption_lower = caption_ids[uuid].lower()
    for keyword in bio_stains:
        if keyword in caption_lower:
            grouped_captions[keyword].append(uuid)

# Print grouped captions
for keyword, group in grouped_captions.items():
    print(f"Captions containing '{keyword}':")
    for caption in group:
        print(f" - {caption}")
    print(f"Number of images with {keyword}: " + str(len(grouped_captions[keyword])) + "\n")
```

```python
# collect all captions that have "tissue" or "cell" in them
# extract the word before it if it's an adjective since it will be describing the type of tissue or cell?
tc = ['tissue', 'tissues', 'cell', 'cells']

grouped_captions = {keyword.lower(): [] for keyword in tc}

# Grouping the captions based on the presence of keywords
for uuid in caption_ids:
    caption_lower = caption_ids[uuid].lower()
    for keyword in tc:
        if keyword in caption_lower:
            grouped_captions[keyword].append(uuid)

# Print grouped captions
for keyword, group in grouped_captions.items():
    print(f"Captions containing '{keyword}':")
    for caption in group:
        print(f" - {caption}")
    print(f"Number of images with {keyword}: " + str(len(grouped_captions[keyword])) + "\n")
```
