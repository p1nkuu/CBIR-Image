# Image Processing with Albumentations

```python
import pandas
import numpy
import random
import matplotlib

import cv2
from matplotlib import pyplot as plt
import albumentations as A
```

Used https://albumentations.ai/docs/getting_started/image_augmentation/ for reference while coding this, and I think it would be helpful to look through the steps as well.

```python
# Reads in image to be augmented
image = cv2.imread(r"C:\Users\tiffa\Documents\bttai\novartis\pubmed_set\images\0a0a7c19-862b-441e-afa7-76295542c576.jpg") # put in path to image here
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # cv2 reads in stuff in BGR but albumentations uses RGB so we have to convert them to RGB

# Display original image
imgplot = plt.imshow(image)
plt.show()
```

```python
# Defines transform function that applies all of the transformations given in A.Compose() as a list
# Check types of transformations that you can do using https://github.com/albumentations-team/albumentations#pixel-level-transforms or https://albumentations.ai/docs/api_reference/augmentations/
transform = A.Compose([
    A.RandomCrop(width=256, height=256),
    A.HorizontalFlip(p=1),
    A.RandomBrightnessContrast(p=1),
])
```

```python
# Apply the transform function on the image to augment it
transformed = transform(image=image)

# transform will return a dictionary with a single key image. Value at that key will contain an augmented image.
transformed_image = transformed["image"]

imgplot_augmented = plt.imshow(transformed_image)
plt.show()
```
