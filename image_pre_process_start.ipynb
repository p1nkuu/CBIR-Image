```python
import pandas as pd
import numpy as np 

from glob import glob 


import cv2
import matplotlib.pylab as plt
```

LOAD PHOTO

```python
test_photo = glob('/Users/genesishang/books_set/images/0add6a3c-ad19-4f9a-8a70-8f86ff5878af.png')
```

```python
img_mp1 = plt.imread(test_photo[0])
img_mp1.shape
```

```python
fig, ax = plt.subplots(figsize=(10,10))
ax.imshow(img_mp1)
ax.axis('off')
plt.show()
```

CONVERT FROM RGBA TO RGB COLORSPACE

```python
img_mp1_rgb = cv2.cvtColor(img_mp1, cv2.COLOR_RGBA2RGB)
fig, ax = plt.subplots(figsize=(8,8))
ax.imshow(img_mp1_rgb)
ax.axis('off')
plt.show()
```

RESIZE IMAGE

```python
img_resized = cv2.resize(img_mp1_rgb, None, fx=0.25, fy=0.25)
fig, ax = plt.subplots(figsize = (8, 8))
ax.imshow(img_resized)
ax.axis('off')
plt.show()
```
