# Image Pre-processing for Matches Dataset

In this notebook, we will:

- Apply advanced image pre-processing techniques to enhance image quality..
- Save the pre-processed images for use in embedding generation
- Clear memory after processing to manage resources efficiently.
---

## **Step 1: Import Libraries**

We begin by importing the necessary libraries.

```python
# Import necessary libraries
import os
import numpy as np
from PIL import Image, ImageEnhance, ImageOps
from tqdm import tqdm
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import gc
```

## **Step 2: Define Helper Functions**

We define functions for image pre-processing.

```python
# Function to apply advanced pre-processing techniques
def advanced_preprocess_image(image_path):
    img = Image.open(image_path).convert('RGB')
    
    # Resize image to 256x256 and center crop to 224x224
    preprocess = transforms.Compose([
        transforms.Resize(256),
        transforms.CenterCrop(224),
        transforms.ColorJitter(brightness=0.1, contrast=0.1, saturation=0.1),
        transforms.ToTensor()
    ])
    img = preprocess(img)
    
    # Convert back to PIL Image
    img = transforms.ToPILImage()(img)
    
    # Apply histogram equalization
    img = ImageOps.equalize(img)
    
    # Optionally, display the pre-processed image
    # plt.imshow(img)
    # plt.axis('off')
    # plt.show()
    
    return img
```

## **Step 3: Load and Preprocess Images**

We load image paths from the matches dataset and apply pre-processing.

```python
# Define matches directory
matches_dir = './matches/'
preprocessed_dir = './matches_preprocessed/'
os.makedirs(preprocessed_dir, exist_ok=True)

image_files = [f for f in os.listdir(matches_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

# Preprocess images and save them
for image_file in tqdm(image_files, desc='Preprocessing Images'):
    image_path = os.path.join(matches_dir, image_file)
    preprocessed_image = advanced_preprocess_image(image_path)
    preprocessed_image.save(os.path.join(preprocessed_dir, image_file))
```

## **Step 4: Clear Memory**

We clear variables and free up memory.

```python
# Clear variables and free memory
del preprocessed_image, image_files
gc.collect()
```

## **Step 5: Conclusion**

We have successfully pre-processed all images in the matches dataset and saved them for embedding generation.
