```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
import json

with open('/Users/genesishang/books_set/captions.json', 'r') as file:
    data = json.loads(file.read())
```

Finding most frequent words

```python

captions = ''

for id in data:
    if data[id]['caption'] not in captions:
        captions += data[id]['caption']

captions = captions.strip()

```

```python
doc = nlp(captions)
```

Filtered list 

```python
filtered_tokens = [token.text for token in doc if not token.is_stop and not token.is_punct]
print(filtered_tokens)
```

Top 50 Words by Occurrence

```python
from collections import Counter

word_freq = Counter(filtered_tokens)
common_words = word_freq.most_common(50)
print(common_words)
```

Top 50 Nouns by Occurrence

```python
nouns = [token.text for token in doc if (not token.is_stop and not token.is_punct and token.pos_ == "NOUN")]

noun_freq = Counter(nouns)
common_nouns = noun_freq.most_common(50)
print(common_nouns)
```

Sorting Captions by Content

```python
import json

with open('/Users/genesishang/books_set/captions.json', 'r') as file:
    data = json.loads(file.read())
```

Organize Caption/UUID dict pairs

```python
captions = {}

for id in data:
    if data[id]['uuid'] not in captions.keys():
        captions[data[id]['uuid']] = data[id]['caption'].strip()

# print(captions)
```

Sort Images by Common Staining Techniques

```python
import pandas as pd

s1 = pd.read_csv('/Users/genesishang/Downloads/BTTAI_ Caption Organization  - Staining Techniques.csv')

staining_techniques = [val[0].lower() for val in s1.values]
# print(staining_techniques)

grouped_captions = {keyword.lower(): [] for keyword in staining_techniques}

# Grouping the captions based on the presence of keywords
for uuid in captions:
    caption_lower = captions[uuid].lower()
    # print(caption_lower)
    for keyword in staining_techniques:
        if keyword in caption_lower:
            grouped_captions[keyword].append(uuid)

# Print grouped captions
for keyword, group in grouped_captions.items():
    print(f"Captions containing '{keyword}':")
    for caption in group:
        print(f" - {caption}")
    print()  # Blank line for readability
```
